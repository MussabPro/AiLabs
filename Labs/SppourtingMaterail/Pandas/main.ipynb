````python
a = 15
b = 8
c = 23
d = 5
e = 12

print(((a+b)*(c-d) + e) / (b-d))
````

```asm
include irvine32.inc

.data
	a dd 15
	b dd 8
	aPb dd ?
	c1 dd 23
	d dd 5
	CminusD dd ?
	e dd 12
	top dd ?
	bMinusD dd ?

.code
main PROC
	mov eax, a
	mov ebx, b
	add eax, ebx
	mov aPb, eax        ; a + b

	mov eax, c1
	mov ebx, d
	sub eax, ebx
	mov CminusD, eax    ; c - d

	mov eax, aPb
	mov ebx, CminusD
	mul ebx             ; EDX:EAX = (a+b)*(c-d)

	add eax, e
	mov top, eax        ; top = (a+b)*(c-d) + e

	mov eax, b
	sub eax, d
	mov bMinusD, eax    ; b - d

	xor edx, edx        ; clear EDX before DIV
	mov eax, top
	mov ebx, bMinusD
	div ebx             ; eax = top / (b-d)

	call WriteInt
	exit
main endp
end main
