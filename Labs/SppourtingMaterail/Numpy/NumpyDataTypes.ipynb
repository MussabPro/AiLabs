```python
binarty_repr = np.binary_repr(arr.max())

print(f"Binary representation of max value: {binarty_repr}")
if len(binarty_repr) < 16:
    arr = arr.astype("int16")
elif len(binarty_repr) < 32:
    arr = arr.astype("int32")
elif len(binarty_repr) < 64:
    arr = arr.astype("int64")
else:
    arr = arr.astype("float64")
print(f"New array dtype: {arr.dtype}")
```

```python
import time
start = time.time()
value = 2**(2**32)
end = time.time()
print("Variable value is 2^",2**32)
bits = value.bit_length()
bytes = bits / 8
kilobytes = bytes / 1024
megabytes = kilobytes / 1024
print("Variable size in MB:", megabytes)
print(f"Time taken: {end - start} seconds")
print(f"Allocating {bits} memory locations for variable. Each having an address.")
```
