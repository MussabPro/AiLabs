```python
import numpy as np
import time

# Python list
size = 1_000_000
list1 = list(range(size))
list2 = list(range(size))

start = time.time()
result = [x + y for x, y in zip(list1, list2)]
end = time.time()
print("Python list addition time:", end - start)

# NumPy array
arr1 = np.array(list1)
arr2 = np.array(list2)

start = time.time()
result = arr1 + arr2  # Vectorized operation
end = time.time()
print("NumPy array addition time:", end - start)
```

```python
import sys

list_data = list(range(size))
numpy_data = np.array(list_data)
# list_data and numpy_data already exist
total_list_size = sys.getsizeof(list_data) + sum(sys.getsizeof(x) for x in list_data)
numpy_size = numpy_data.nbytes

print("Approx. size of Python list in MB:", total_list_size / 1024**2)
print("Size of NumPy array in MB:", numpy_size / 1024**2)
```

```python
import numpy as np
import time

# Python list
size = 1_000_000
list1 = list(range(size))


start = time.time()
result = [x**2 for x in list1]
end = time.time()
print("Python list squaring time:", end - start)

# NumPy array
arr1 = np.array(list1)


start = time.time()
result = arr1**2  # Vectorized operation
end = time.time()
print("NumPy array squaring time:", end - start)
```
