```

        v

```

## Step 1: Define the Email Templates

**What this block does:** We create a dictionary called `email_templates_by_tone`. It stores three pre-written email templates, one for each tone: formal, friendly, and urgent.

**Why we need it:** Instead of writing a new email from scratch every time, the agent looks up the right template for the requested tone. This is how simple AI writing assistants work - they start with a structure and fill in the blanks.

**What to expect as output:** When you run this cell, Python will store the templates in memory. You will see a confirmation message and a list of the available tones.

````python
# This dictionary holds one email template per tone.
# Each template uses {placeholder} marks that we will fill in later.
# Key = tone name (string), Value = template text (string)
email_templates_by_tone = {

    'formal': (
        'Subject: {subject}\n'
        '\n'
        'Dear {recipient_name},\n'
        '\n'
        'I hope this message finds you well. I am writing to you in your capacity '
        'as {recipient_role} regarding the following matter: {purpose}.\n'
        '\n'
        'I would be grateful if you could review this at your earliest convenience '
        'and provide any guidance or feedback you may have. Please do not hesitate '
        'to contact me if you require any additional information.\n'
        '\n'
        'Thank you for your time and consideration.\n'
        '\n'
        'Yours sincerely,\n'
        '{sender_name}'
    ),

    'friendly': (
        'Subject: {subject}\n'
        '\n'
        'Hi {recipient_name},\n'
        '\n'
        'Hope you are doing great! I wanted to reach out to you as {recipient_role} '
        'about something I think you will find interesting: {purpose}.\n'
        '\n'
        'Let me know what you think when you get a chance - I am really looking '
        'forward to hearing your thoughts. Feel free to reply whenever it suits you.\n'
        '\n'
        'Cheers,\n'
        '{sender_name}'
    ),

    'urgent': (
        'Subject: URGENT - {subject}\n'
        '\n'
        'Dear {recipient_name},\n'
        '\n'
        'This is a time-sensitive matter that requires your immediate attention. '
        'As {recipient_role}, you are the right person to address this: {purpose}.\n'
        '\n'
        'Please respond as soon as possible - ideally within the next few hours. '
        'Delay in addressing this matter may have significant consequences. '
        'I am available at any time to discuss this further.\n'
        '\n'
        'With urgency,\n'
        '{sender_name}'
    )
}

# Print the available tone options so students can see what the agent supports
print('Email templates loaded successfully.')
print(f'Available tones: {list(email_templates_by_tone.keys())}')
````

---
## Step 2: Build the Simulated AI Function (fake_llm)

**What this block does:** We define a function called `fake_llm()`. In a real AI system, this function would send your request to a language model and get back a generated response. In our offline lab, it simply picks the right template from our dictionary and fills in the placeholders.

**Why we need it:** This function is the 'brain' of our agent. By wrapping the template logic inside `fake_llm()`, we make it easy to swap in a real AI model later without changing the rest of the code.

**What to expect as output:** Running this cell defines the function. You will see a confirmation message.

````python
# SIMULATED AI: In a real system this would call a language model API.
# Here, we pick a template based on tone and fill in the details.

def fake_llm(recipient_name, recipient_role, email_purpose, email_tone, sender_name):
    """
    Simulates an AI language model that writes a personalized email.

    Takes in:
      recipient_name  - the name of the person receiving the email (string)
      recipient_role  - the job title or role of the recipient (string)
      email_purpose   - a short description of why we are writing (string)
      email_tone      - must be 'formal', 'friendly', or 'urgent' (string)
      sender_name     - the name of the person sending the email (string)

    Returns:
      A completed email as a single string, or an error message if the
      tone is not recognized.
    """

    # Check that the requested tone actually exists in our template dictionary
    if email_tone not in email_templates_by_tone:
        available_tones = list(email_templates_by_tone.keys())
        return f"Error: tone '{email_tone}' is not supported. Choose from: {available_tones}"

    # Retrieve the correct template for the given tone
    selected_template = email_templates_by_tone[email_tone]

    # Build a short subject line based on the purpose
    email_subject = email_purpose[:50] if len(email_purpose) > 50 else email_purpose

    # Fill in all the placeholders in the template with real values
    completed_email_text = selected_template.format(
        recipient_name=recipient_name,
        recipient_role=recipient_role,
        purpose=email_purpose,
        subject=email_subject,
        sender_name=sender_name
    )

    return completed_email_text


print('fake_llm() function defined successfully.')
print('This function acts as our simulated AI brain for this lab.')
````

---
## Step 3: Build the Email Writer Agent

**What this block does:** We define the main agent function called `email_writer_agent()`. This function acts as the coordinator - it receives the request, calls `fake_llm()` to generate the email, adds a date header, and prints the result in a nicely formatted way.

**Why we need it:** In agent design, the agent function is the entry point. It handles the user's request from start to finish, calling tools (like `fake_llm`) along the way. This separation of the agent from the AI brain is an important design pattern.

**What to expect as output:** This cell just defines the function. No email is printed yet.

````python
def email_writer_agent(recipient_name, recipient_role, email_purpose, email_tone, sender_name):
    """
    The main Email Writer Agent.

    Takes in:
      recipient_name  - the name of the person to write to (string)
      recipient_role  - the job title of the recipient (string)
      email_purpose   - what the email is about (string)
      email_tone      - 'formal', 'friendly', or 'urgent' (string)
      sender_name     - the name of the person writing the email (string)

    Returns:
      Nothing - it prints the email directly to the screen.
    """

    # Get today's date to add to the top of the email
    todays_date = datetime.date.today().strftime('%B %d, %Y')

    # Print the agent's thinking step so students can see the decision process
    print('=' * 60)
    print('AGENT THINKING:')
    print(f'  Recipient : {recipient_name} ({recipient_role})')
    print(f'  Purpose   : {email_purpose}')
    print(f'  Tone      : {email_tone}')
    print(f"  Selecting template for tone: '{email_tone}'...")
    print('=' * 60)

    # Call the simulated AI to generate the email body
    generated_email_text = fake_llm(
        recipient_name=recipient_name,
        recipient_role=recipient_role,
        email_purpose=email_purpose,
        email_tone=email_tone,
        sender_name=sender_name
    )

    # Add the date header above the generated email
    full_email_with_header = f'Date: {todays_date}\n{generated_email_text}'

    # Wrap lines to 70 characters so the email looks neat in the terminal
    wrapped_email_lines = []
    for single_line in full_email_with_header.split('\n'):
        wrapped_email_lines.append(textwrap.fill(single_line, width=70))
    final_formatted_email = '\n'.join(wrapped_email_lines)

    # Print the completed email
    print('GENERATED EMAIL:')
    print('-' * 60)
    print(final_formatted_email)
    print('-' * 60)
    print('Agent finished writing the email.')
    print()


print('email_writer_agent() function defined successfully.')
````

---
## Step 4: Prepare the Sample Email Requests

**What this block does:** We create a list called `list_of_email_requests`. Each item in the list is a dictionary that describes one email request. We have three requests with three different tones.

**Why we need it:** Rather than calling the agent three times manually, we store all requests in a list and loop over them. This shows how an agent can handle a batch of tasks automatically.

**What to expect as output:** A confirmation message showing how many requests are queued.

````python
# Each dictionary in this list is one email request for the agent to handle.
# Keys match the parameter names of email_writer_agent() exactly.
list_of_email_requests = [
    {
        'recipient_name': 'Dr. Sarah Mitchell',
        'recipient_role': 'Head of the Computer Science Department',
        'email_purpose': 'requesting approval to present my final year project at the annual tech conference',
        'email_tone': 'formal',
        'sender_name': 'Ahmed Al-Rashid'
    },
    {
        'recipient_name': 'James',
        'recipient_role': 'lab partner and fellow student',
        'email_purpose': 'reminding him about our group study session this Thursday at 4 PM in the library',
        'email_tone': 'friendly',
        'sender_name': 'Fatima'
    },
    {
        'recipient_name': 'Mr. David Chen',
        'recipient_role': 'IT Support Manager',
        'email_purpose': 'reporting that the lab computers have stopped working and we have an exam in 2 hours',
        'email_tone': 'urgent',
        'sender_name': 'Prof. Leila Ahmadi'
    }
]

print(f'Email requests loaded: {len(list_of_email_requests)} emails queued for the agent.')
````

---
## Step 5: Run the Agent on All Three Requests

**What this block does:** We loop over `list_of_email_requests` and call `email_writer_agent()` for each one. The agent will print its thinking and the generated email for each request.

**Why we need it:** This is the main demo run. You will see all three emails written in different tones showing how the same agent handles different situations.

**What to expect as output:** Three complete emails printed one after another, each preceded by the agent's thinking log.

````python
# Loop over each email request and run the agent
for request_number, single_email_request in enumerate(list_of_email_requests, start=1):
    print(f'\n*** EMAIL {request_number} of {len(list_of_email_requests)} ***')

    email_writer_agent(
        recipient_name=single_email_request['recipient_name'],
        recipient_role=single_email_request['recipient_role'],
        email_purpose=single_email_request['email_purpose'],
        email_tone=single_email_request['email_tone'],
        sender_name=single_email_request['sender_name']
    )
````

**What to expect as output:** An error message from the agent explaining that the tone is not supported, but no Python crash or exception.
