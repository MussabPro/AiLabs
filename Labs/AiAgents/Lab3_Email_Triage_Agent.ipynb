## Step 1: Define the Keyword Rules

**What this block does:** We define three lists of keywords. If any urgent keyword appears in the email, it is Urgent. If any spam keyword appears, it is Spam. Otherwise it is Normal. Urgent is checked first because it takes higher priority.

**Why we need it:** The keyword lists are the 'brain' of our classifier. Changing these lists changes how the agent behaves - just like retraining a real classifier.

**What to expect as output:** The three keyword lists are printed.

````python
# Keywords that signal an email is Urgent
# These are checked against subject and body text (case-insensitive)
list_of_urgent_keywords = [
    'urgent', 'asap', 'immediately', 'critical', 'emergency',
    'deadline today', 'action required', 'final notice'
]

# Keywords that signal an email is Spam
list_of_spam_keywords = [
    'winner', 'free gift', 'click here', 'unsubscribe', 'limited offer',
    'congratulations', 'you have been selected', 'earn money', 'prize'
]

# The order of checks: Urgent is highest priority, then Spam, then Normal
# This variable is just for display and documentation
classification_priority_order = ['Urgent', 'Spam', 'Normal']

print('Keyword rules loaded.')
print(f'Urgent keywords ({len(list_of_urgent_keywords)}): {list_of_urgent_keywords}')
print(f'Spam keywords ({len(list_of_spam_keywords)}): {list_of_spam_keywords}')
````

---
## Step 2: Build the Classifier Function

**What this block does:** We define `classify_email()`. It combines the subject and body of an email into one string, then searches that string for urgent keywords, then spam keywords. It returns a label: 'Urgent', 'Spam', or 'Normal'.

**Why we need it:** This is the core intelligence of the triage agent. By using `re.search()` with `re.IGNORECASE`, we match keywords regardless of capitalisation.

**What to expect as output:** The function is defined and a quick test runs.

````python
# SIMULATED AI: In a real system this would use a trained machine learning model.
# Here, we use keyword rules to classify each email.

def classify_email(email_subject, email_body):
    """
    Classifies an email as 'Urgent', 'Spam', or 'Normal'.

    Uses keyword rules checked against the combined subject and body text.
    Urgent keywords are checked first because they have higher priority.

    Takes in:
      email_subject - the subject line of the email (string)
      email_body    - the main text of the email (string)

    Returns:
      A string: 'Urgent', 'Spam', or 'Normal'
    """

    # Combine subject and body into one string to search both at once
    combined_email_text = email_subject + ' ' + email_body

    # Check for urgent keywords first (highest priority)
    for keyword in list_of_urgent_keywords:
        if re.search(keyword, combined_email_text, re.IGNORECASE):
            return 'Urgent'

    # Check for spam keywords next
    for keyword in list_of_spam_keywords:
        if re.search(keyword, combined_email_text, re.IGNORECASE):
            return 'Spam'

    # No keyword matched - classify as Normal
    return 'Normal'


# Quick test to verify the classifier works before the full demo
print('Classifier test results:')
print(f'  URGENT subject -> {classify_email("URGENT: server down", "please fix this asap")}')
print(f'  Spam subject   -> {classify_email("You are a winner!", "Click here for your free gift")}')
print(f'  Normal subject -> {classify_email("Team lunch Friday", "See you at noon.")}')
````

## Step 4: Build the Triage Agent

**What this block does:** We define `inbox_triage_agent()`. It loops over all the emails in the inbox, calls `classify_email()` for each one, and adds a `'label'` key to each email dictionary with the classification result.

**Why we need it:** Adding the label directly to the dictionary keeps the email and its classification together in one place - this makes sorting and reporting easier.

**What to expect as output:** The function is defined. A confirmation message is printed.

````python
def inbox_triage_agent(list_of_emails):
    """
    The main Inbox Triage Agent.

    Loops over a list of email dictionaries, classifies each one,
    and adds a 'label' key ('Urgent', 'Spam', or 'Normal') to each.

    Takes in:
      list_of_emails - a list of email dictionaries, each with
                       'sender', 'subject', and 'body' keys

    Returns:
      The same list with a 'label' key added to every dictionary.
    """

    print('Agent is triaging the inbox...')

    # Process each email one by one
    for single_email in list_of_emails:
        # Classify the email using the keyword rules
        classification_result = classify_email(
            email_subject=single_email['subject'],
            email_body=single_email['body']
        )

        # Add the label directly to the email dictionary
        single_email['label'] = classification_result

    print(f'Done. {len(list_of_emails)} emails classified.')
    return list_of_emails


print('inbox_triage_agent() defined successfully.')
````

---
## Step 5: Run the Triage Agent

**What this block does:** We call `inbox_triage_agent()` on our sample inbox. After it runs, every email dictionary will have a `'label'` key.

**Why we need it:** This is where the agent does its work - reading every email and labelling it.

**What to expect as output:** A message saying how many emails were classified.

````python
# Run the triage agent on the sample inbox
# After this call, each email dictionary will have a 'label' key
classified_inbox = inbox_triage_agent(list_of_sample_inbox_emails)

# Peek at the first email to confirm the label was added
print('\nFirst email after classification:')
first_email = classified_inbox[0]
print(f"  Sender : {first_email['sender']}")
print(f"  Subject: {first_email['subject']}")
print(f"  Label  : {first_email['label']}")
````

---
## Step 6: Sort and Display the Classified Inbox

**What this block does:** We sort the emails so Urgent comes first, then Normal, then Spam. We use a custom sort key that assigns a number to each label (Urgent=0, Normal=1, Spam=2). Then we print the sorted inbox in a readable table format.

**Why we need it:** A sorted inbox is much more useful than a random one. This is how real email clients present your inbox - important messages at the top.

**What to expect as output:** A formatted inbox table sorted by priority.

````python
# Assign a sort number to each label so we can sort the list
# Lower number = higher in the inbox (Urgent first)
label_sort_order = {'Urgent': 0, 'Normal': 1, 'Spam': 2}

# Sort the inbox list by the label's sort number
sorted_inbox_emails = sorted(
    classified_inbox,
    key=lambda email_dict: label_sort_order[email_dict['label']]
)

# Print the sorted inbox as a formatted table
print('=' * 72)
print('SORTED INBOX')
print('=' * 72)
print(f'{"#":<4} {"Label":<8} {"From":<30} {"Subject"}')
print('-' * 72)

for email_index, email_item in enumerate(sorted_inbox_emails, start=1):
    truncated_subject = email_item['subject'][:35]
    truncated_sender = email_item['sender'][:28]
    print(f"{email_index:<4} {email_item['label']:<8} {truncated_sender:<30} {truncated_subject}")

print('=' * 72)
````

---
## Step 7: Print Category Counts

**What this block does:** We count how many emails fall into each category and print a small summary. This is the final report the triage agent delivers to the user.

**Why we need it:** A summary count gives a quick overview of the inbox health - how many urgent items need attention, how much is spam, and how much is normal.

**What to expect as output:** Three counts and a total.

````python
# Count emails in each category
count_of_urgent_emails = sum(1 for email in classified_inbox if email['label'] == 'Urgent')
count_of_spam_emails = sum(1 for email in classified_inbox if email['label'] == 'Spam')
count_of_normal_emails = sum(1 for email in classified_inbox if email['label'] == 'Normal')

print('CATEGORY COUNTS')
print('-' * 30)
print(f'  Urgent : {count_of_urgent_emails} email(s)')
print(f'  Normal : {count_of_normal_emails} email(s)')
print(f'  Spam   : {count_of_spam_emails} email(s)')
print('-' * 30)
print(f'  Total  : {len(classified_inbox)} email(s)')
````
