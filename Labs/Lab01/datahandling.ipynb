## 1. Dictionaries, tuples, and lists
A dictionary stores named values. The position is a tuple, while the sensors value is a list.

```python
agent_state = {
    'position': (2, 3),
    'energy': 87,
    'sensors': ['bump', 'clear', 'clear', 'bump']
}

print('Agent state:', agent_state)
print('Energy level:', agent_state['energy'])
```

## 2. NumPy arrays
NumPy arrays support fast numerical operations. The normalization expression changes every reading to a value between 0 and 1.

```python
import numpy as np

readings = np.array([12.5, 9.3, 15.7, 20.1, 8.4])

print('All readings:', readings)
print('Average:', readings.mean())
print('Highest:', readings.max())
print('Lowest:', readings.min())

normalized = (readings - readings.min()) / (readings.max() - readings.min())
print('Normalized:', normalized)
```

## 3. Pandas DataFrames
A DataFrame is a table in Python. This example fills a missing value and filters the table to keep only fast trials.

```python
import pandas as pd

df = pd.DataFrame({
    'trial': [1, 2, 3, 4, 5],
    'success': [1, 0, 1, 1, None],
    'time_sec': [3.2, 5.1, 2.8, 4.0, 3.9]
})

print(df)
print()
print('Summary statistics:')
print(df.describe())

df['success'] = df['success'].fillna(0)
fast_trials = df[df['time_sec'] < 4.0]

print()
print('Fast trials:')
print(fast_trials)
```

## 4. Draw a histogram
This cell saves a histogram of the sensor readings as `lab1_histogram.png`. Saving the figure works even when the notebook has no display window.

```python
import matplotlib.pyplot as plt

plt.hist(readings, bins=5)
plt.title('Sensor Reading Distribution')
plt.xlabel('Reading value')
plt.ylabel('How many times it occurred')
plt.savefig('lab1_histogram.png')
plt.show()

print('Chart saved as lab1_histogram.png')
```
